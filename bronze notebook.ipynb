```python
customers = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("abfss://inputdata@storagelake9720.dfs.core.windows.net/raw/customers_bad.csv")

orders = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("abfss://inputdata@storagelake9720.dfs.core.windows.net/raw/orders_bad.csv")


products = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("abfss://inputdata@storagelake9720.dfs.core.windows.net/raw/products_bad.csv")
```

```python
from pyspark.sql.functions import *
customers_bronze = (
    customers
    .withColumn("load_timestamp", current_timestamp())
    .withColumn("source_file_name", lit('_metadata.file_path'))
)
customers_bronze.write.format("delta").mode("overwrite").save("abfss://inputdata@storagelake9720.dfs.core.windows.net/bronze/customers")
```

```python
from pyspark.sql.functions import *
orders_bronze = (
    orders
    .withColumn("load_timestamp", current_timestamp())
    .withColumn("source_file_name", lit('_metadata.file_path'))
)
orders_bronze.write.format("delta").mode("overwrite").option("overwriteSchema", "true").save("abfss://inputdata@storagelake9720.dfs.core.windows.net/bronze/orders")
```

```python
from pyspark.sql.functions import *
product_bronze = (
    products
    .withColumn("load_timestamp", current_timestamp())
    .withColumn("source_file_name", lit('_metadata.file_path'))
)
product_bronze.write.format("delta").mode("overwrite").option("overwriteSchema", "true").save("abfss://inputdata@storagelake9720.dfs.core.windows.net/bronze/products")
```
